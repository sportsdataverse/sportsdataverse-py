# Women's basketball intro — sportsdataverse-py

A combined tour of the NCAA `wbb` and pro `wnba` submodules — they share a near-parallel `espn_wbb_*` / `espn_wnba_*` surface, so once you know one you know the other.

R companion: [wehoop](https://wehoop.sportsdataverse.org). Part of the [SportsDataverse](https://py.sportsdataverse.org/docs/ecosystem).

````python
import polars as pl
import sportsdataverse as sdv
````

## Teams (parallel WNBA / WBB calls)

````python
wnba_teams = sdv.wnba.espn_wnba_teams()
wnba_teams.shape
````

````python
wbb_teams = sdv.wbb.espn_wbb_teams()
wbb_teams.shape
````

## Team roster — UConn (WBB) and Las Vegas Aces (WNBA)

````python
uconn_roster = sdv.wbb.espn_wbb_team_roster(team_id=2509, season=2025)
uconn_roster.select(['athlete_id', 'athlete_display_name', 'jersey', 'position_abbreviation']).head()
````

````python
aces_roster = sdv.wnba.espn_wnba_team_roster(team_id=17, season=2024)
aces_roster.select(['athlete_id', 'athlete_display_name', 'jersey', 'position_abbreviation']).head()
````

## Schedule

````python
wnba_sched = sdv.wnba.espn_wnba_schedule(dates=20241020)
wnba_sched.select(['id', 'home_team_full_name', 'away_team_full_name', 'home_score', 'away_score']).head()
````

## PBP — 2024 WNBA Finals Game 5

`espn_wnba_pbp` returns a dict of component frames; `plays` is the main one.

````python
pbp = sdv.wnba.espn_wnba_pbp(game_id=401620238)
list(pbp.keys())[:8]
````

````python
plays = pl.from_pandas(pbp['plays'])
plays.select(['period_number', 'clock_display_value', 'text', 'scoring_play']).head()
````

## Season player + team stats — multi-table dict shape

ESPN groups stats by category (`general`, `offensive`, `defensive`, ...). The wrappers flatten that into a single tidy long frame.

````python
# WBB: Paige Bueckers (UConn) — athlete_id 4433402, season 2025
pb_stats = sdv.wbb.espn_wbb_player_stats(athlete_id=4433402, season=2025)
pb_stats.shape
````

````python
pb_stats.head()
````

````python
# WNBA: Caitlin Clark — athlete_id 4433403, season 2024
cc_stats = sdv.wnba.espn_wnba_player_stats(athlete_id=4433403, season=2024)
cc_stats.head()
````

````python
uconn_team_stats = sdv.wbb.espn_wbb_team_stats(team_id=41, season=2025)
uconn_team_stats.head()
````

## Standings

````python
wnba_standings = sdv.wnba.espn_wnba_standings(season=2024)
wnba_standings.select(['team_id', 'team_display_name', 'wins', 'losses', 'win_percent']).head()
````

````python
wbb_standings = sdv.wbb.espn_wbb_standings(season=2025)
wbb_standings.select(['team_id', 'team_display_name', 'wins', 'losses', 'win_percent']).head()
````

## Draft (WNBA)

````python
draft_2024 = sdv.wnba.espn_wnba_draft(season=2024)
draft_2024.select(['overall_pick', 'team_display_name', 'athlete_display_name', 'athlete_position_abbreviation']).head()
````
